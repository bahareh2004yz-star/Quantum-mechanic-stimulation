```python
import numpy as np
import matplotlib.pyplot as plt
# Length of the box in meters 
L = 1.0
# Reduced Planck's constant (J·s)
hbar = 1.0545718e-34
# Mass of the particle (electron mass in kg)
m = 9.10938356e-31

def psi_n(n, x, L):
    return np.sqrt(2 / L)* np.sin(n* np.pi * x / L) 

def energy_n(n, L, m):
    return (n**2 * np.pi**2 * hbar**2) / (2 * m * L**2)


def plot_wave_function(n_values, L):
    x = np.linspace(0, L, 1000)

    plt.figure(figsize = (10,6))

    for n in n_values:
        psi = psi_n(n, x, L)
        plt.plot(x, psi, label = f"ψ_{n}(x)", lw=2) 

    plt.title("Wave Functions for Particles in a box ") 
    plt.xlabel("position (x)")   
    plt.ylabel("ψ_n(x)")
    plt.grid(True)
    plt.legend()
    plt.show()
# Function to plot probability densities
#         
def plot_prob_density(n_values, L):
        x = np.linspace(0, L, 1000)
        plt.figure(figsize = (10,6))


        for n in n_values:
            psi = psi_n(n, x, L)  
            prob_density = psi**2
            plt.plot(x, prob_density, label =f"|ψ_{n}(x)|^2")

        plt.title("Wave Functions for Particles in a box ") 
        plt.xlabel("position (x)")   
        plt.ylabel("ψ_n(x)|^2")
        plt.grid(True)
        plt.legend()
        plt.show()

def energy_levels(n_values, L, m):
     print("Energy levels for different quantum numbers (in Joules):")
     for n in n_values:
          E = energy_n(n, L, m)
          print(f"E_{n} = {E: 2e} J")
# Example

n_values = [1, 2, 3, 4, 5]
plot_wave_function(n_values,L)
plot_prob_density(n_values, L)
energy_levels(n_values,L, m)


```
